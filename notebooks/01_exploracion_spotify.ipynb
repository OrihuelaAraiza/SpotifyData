# Fase 1: Exploracion y comprension

Identifica estructura, JSON principales, contenido y riesgos de privacidad de ambos historiales de Spotify.


```python
from pathlib import Path
import json
base = Path('..')
for folder in ['Spotify Extended Streaming History','Spotify Extended Streaming History 2']:
    files = sorted((base / folder).glob('Streaming_History_Audio_*.json'))
    video = sorted((base / folder).glob('Streaming_History_Video_*.json'))
    rows = sum(len(json.load(open(path, encoding='utf-8'))) for path in files)
    print(folder, 'audio_json=', len(files), 'video_json=', len(video), 'filas_audio=', rows)
```

Reflexion: el historial permite inferir horarios, dispositivos, paises de conexion, gustos persistentes y cambios de rutina. Por eso el proyecto exporta resultados agregados y no direcciones IP.
